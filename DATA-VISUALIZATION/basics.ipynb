```python
import numpy as np
import pandas as pd
import  matplotlib.pyplot as plt
```

```python
x=np.linspace(0,6,8).round(1)
x
```

```python
y=x**3
y
```

```python
plot=plt.plot(x,y, marker='o', markersize=6, color='blue')
plt.title("Example")
plt.xlabel("xlabel")
plt.ylabel("ylabel")

# 4. Display the grid (optional, but helps see coordinates)
plt.grid(True)

# 5. Show the plot
plt.show()
```
