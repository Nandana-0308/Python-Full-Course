### Memory & Copy Behavior: Python Lists vs. NumPy Arrays

| Operation Type | Standard Python (`list`) | NumPy (`np.array`) |
| :--- | :--- | :--- |
| **Assignment** <br>`y = x` | **Shared Object:** Both variables point to the exact same list. Modifying `y` changes `x`. | **Shared Object:** Both variables point to the exact same array object. Modifying `y` changes `x`. |
| **Shallow Copy** <br>Python: `x.copy()` <br>NumPy: `x.view()` | **New Box, Shared Inside:** Copies the outer list, but shares inner mutable objects. Modifying a flat number won't affect `x`, but modifying a nested list **will** affect `x`. | **Shared Memory (View):** Creates a new window to the exact same block of numbers. Modifying *any* value in `y` **immediately changes** `x`. |
| **Deep Copy** <br>Python: `deepcopy(x)` <br>NumPy: `x.copy()` | **Complete Disconnection:** Recursively crawls and duplicates every single element. They are 100% independent. | **Complete Disconnection:** Instantly duplicates the entire raw, flat block of data into a brand new memory slot. They are 100% independent. |
| **Slicing** <br>`y = x[1:4]` | Creates a **Shallow Copy**. | Creates a **View (Shallow Copy)** automatically to save memory performance. |


### in std python

```python
a=[[1,2,3,4,5],[6,7,8,9,0]]
b=a

print(a)
print(b)
print(a==b)  # values inside list a and list b are exactly the same ,ref same ,
# • It returns True if both lists have the exact same elements in the exact same order.
# • It returns False if any element or the order is different.

print(a is b)   # Checks if they point to the exact same block in memory

a[1][3] == a[1][3]   # 9==9 true

c=a.copy()


print(c==a)
print(a is c)
```

```python
import copy

#shallow
a=[[11,21,13,41,15],[56,47,38,29,10]]
b=copy.copy(a)
print(a==b)
print(a is b)

a[1][3]=45

print(a)
print(b)

#__________________________________________________________________

#deeep
p=[[34,54,67,54,22,57,97],[43,21,23,1,246,755,43]]
q=copy.deepcopy(p)
print(q==p)
print(q is p)

q[0][4]=100

print(p)
print(q)
```
