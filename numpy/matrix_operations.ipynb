# Matrix Operations

```python
import numpy as np
```

```python
A = np.array([[1, 2], [4, 5]])
print(A)
print("-" * 50)
B = np.array([[11, 21], [45, 56]])
print(B)

print("-" * 50)

print(f"Matrix Multiplication A x B :  \n {A @ B}")
print("-" * 50)

print(np.dot(A,B))
print("-" * 50)

print(A.T)
```
