## Lists and NumPy Arrays

Python lists and NumPy arrays behave differently during multiplication.


```python
import numpy as np

# Check NumPy version
print(np.__version__) 
```

### Doubling list elements using a loop

```python

numbers = [1, 2, 3, 4]
doubled = []

for x in numbers:
    doubled.append(x * 2)

print(doubled)
```

Using list comprehension

```python
numbers = [1, 2, 3, 4]

doubled = [x * 2 for x in numbers]

print(doubled)
```

Using map() and lambda

```python
numbers = [1, 2, 3, 4]

doubled = list(map(lambda x: x * 2, numbers))

print(doubled)
```

Using a NumPy array

```python
array = np.array([1, 2, 3, 4])

array = array * 2

print(array)
```

### Dimention of array ---> array_name.ndim

```python
array=np.array('A')  #0 - D array ,, np.array(['A'])--- 1D array
print(array)
print(array.ndim)
```

```python

# 1. np.array('a')

# By passing a single string literal, you create a 0-D scalar array. In NumPy, this represents a single standalone value without any axis.
# • It holds exactly one value, but it has no length or direction.
# • Think of it as a wrapper around a Python scalar value that allows it to use NumPy functions.

# 2. np.array(['a'])

# By passing a Python list containing a string ['a'], you create a 1-D array.
# • This array has exactly one axis (Axis 0).
# • Even though it only contains one element, it acts as a collection/sequence. You can iterate over it or pull elements out by their index positions.
```
