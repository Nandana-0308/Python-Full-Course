Random Numbers

```python
import numpy as np
rng = np.random.default_rng()  #default_rng(seed=1) ----> produce the same result as before  
#rng= random number generator 
```

```python

print(rng.integers(1,7)) 
#in keyword argument integers(low=3 high=50 size=3)
```

```python
#random array generation 
print("Random array")
print(rng.integers(low=1,high=7,size=6)) 
print('-'*50)

print("Random Matrix 2D ")
#random 2d matrix
print(rng.integers(1,7,(3,5))) 
print('-'*50)


print("Random Matrix 3D ")
#random 2d matrix
print(rng.integers(1,7,(3,2,4))) 
print('-'*50)

```

```python
print(np.random.choice(100))
print(np.random.choice([23,45,3,21,544,6,44]))
print(np.random.choice([3, 5, 7, 9], size=(3, 5)))
```

### Shuffle

```python
array=np.array([1,2,3,4,5])
rng.shuffle(array)
print(array)
```

```python
fruits=np.array(["🍋‍🟩","🍉","🍋","🍎","🍌","🍐","🍏","🍇"])
print(fruits)
print('-'*50,"\n")
fruits=rng.choice(fruits,size=(3,4))
print(fruits)
```
