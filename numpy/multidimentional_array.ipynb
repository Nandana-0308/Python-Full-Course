## ACESSING ARRAY ELEMENTS


```python
arr_1d = np.array([10, 20, 30, 40, 50])

print(f" 4th element (index is 3) : {arr_1d[3]}")   # Output: 40
print(f"last element : {arr_1d[-1]}")
```

```python
arr_2d = np.array([[10, 20, 30], 
                   [35, 40, 45], 
                   [34, 56, 78]])

print(arr_2d[0, 2])  # Output: 15
print(arr_2d[-1, -1])  # last row last column
```

```python
arr_3d = np.array([
    [[1, 2, 3], [4, 5, 6]],     # Layer 0
    [[10, 20, 30], [40, 50, 60]] # Layer 1
])
print(arr_3d[0, 1, 1])  # Output: 5   [layer, row, column]
print(arr_3d[1, 0, 2])
print(arr_3d[0])   #Giving only the layer)
print(arr_3d[0, 1]) #Giving layer and row

```

### Diff bw comma (,) and seperator (:)

In NumPy, the comma (,) and the colon (:) serve completely different purposes when indexing and slicing arrays.
The short explanation: The comma navigates between different dimensions (axes), while the colon selects a range of elements within a single dimension.

1. The Comma (,) — Dimension Separator

The comma is used to move across dimensions. In a 2D array, it separates your Row selection from your Column selection.
• Syntax: array[row_index, column_index]
• Think of it as: Coordinates on a map (X, Y).
python
import numpy as np

<!-- # Let's use a simple 2D array
# Row 0: [10, 20, 30]
# Row 1: [40, 50, 60] -->
matrix = np.array([[10, 20, 30], [40, 50, 60]])

<!-- # Grab the element at Row 1, Column 2 -->
print(matrix[1, 2])  # Output: 60
Use code with caution.

2. The Colon (:) — Range Slicer

The colon is used to extract a subset or a range of data along a single dimension. It means "give me everything from here to there".
• Syntax: array[start:end]
• Think of it as: A cutting tool to slice a chunk out of a sequence.
python
<!-- # Grab elements from index 0 up to (but excluding) index 2
# (Using a 1D array example for simplicity) -->
vector = np.array([10, 20, 30, 40, 50])
print(vector[0:2])  # Output: [10, 20]

<!-- # A single colon [ : ] by itself means "select the entire axis" -->
Use code with caution.

Direct Comparison (Using them together)

When you combine commas and colons, you can pull out entire rows, columns, or sub-grids from a multi-dimensional array.

Code	Meaning	What it extracts
matrix[1, 2]---- Specific Row , Specific Column	------ A single scalar value (60)
matrix[0, :] ----	Row 0 , All Columns ---	The entire first row ([10, 20, 30])
matrix[:, 1]---- All Rows , Column 1--- The entire second column ([20, 50])
matrix[0:2, 0:2] ---	Rows 0 to 1 , Columns 0 to 1---- A sub-grid matrix ([[10, 20], [40, 50]])


```python
arr1=np.array([1,2,3,4,5,6,7,8])

print(arr1)

print('-'*50)

# print(arr1[0,4])  IndexError   -- no rows and columns 
print(arr1[1:4]) #[2 3 4]
print('-'*50)
print(arr1[2::2]) #[3 5 7]
```

```python
arr2 = np.array([[1, 2, 3], [4, 5, 6], [8, 9, 0], [21, 34, 22], [67, 87, 98]])
print(arr2)
print("-" * 50)
# print(arr2[1,3])  #row 1 coulmn 3 --error
print(arr2[1,2]) #row 1 column 2 --- 6
print("-" * 50)

print(arr2[0:4])   #[1:23] also possible ,no error , because it is end 
print("-" * 50)

print(arr2[:4:2])   #element in index 0, 2 , pritied , 4th excluded becasue it is in the 4th index -- ie we provided 4 as the end 
print("-" * 50)

print(arr2[:10:2])  # now  4th will be pritined 
```
