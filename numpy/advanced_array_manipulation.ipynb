# Advanced array manipulation


### stacking array

marix cant be stack , arrays  can stack

```python
import numpy as np
```

```python
a=np.arange(11,16)
b=np.arange(21,26)
print(a,b)

print("-"*50)

print(np.vstack((a,b)))
print("-"*50)

print(np.hstack((a,b)))
print("-"*50)

print(np.column_stack((a,b)))
```

### Splitting

```python
a=np.arange(16).reshape(4,4)
print(a)
print("-"*50)

print(np.hsplit(a,4))
print("-"*50)

print(np.vsplit(a,2))
```
