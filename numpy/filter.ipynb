# Filtering

```python
import numpy as np
```

```python
ages = np.array([[21, 23, 42, 76, 71, 16, 45], 
                 [45, 92, 12, 34, 17, 8, 66]])

teenagers = ages[ages < 18]
print(teenagers)

adults =ages[(ages>=18) & (ages<65)]
print(adults)

even_agse=ages[ages%2==0]
odd_ages =ages[ages%2!=0] 
print(even_agse)
print(odd_ages)
```

### where(condition,array,fill_value)

```python
adults = np.where(ages >= 18, ages, 0)
print(adults)
```

```python
arr = np.array([1, 12, 15, 18, 21, 24])

# Filter elements that are greater than 10 AND divisible by 3
mask = (arr > 10) & (arr % 3 == 0)

print(arr[mask])
# Output: [12 15 18 21 24]
```
