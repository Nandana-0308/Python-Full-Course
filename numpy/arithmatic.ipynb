# Array Arithmatic


```python
import numpy as np
```

### Scalar Arithmatic
single value

```python
arr = np.array([1, 2, 3, 4])

print(arr + 2)
print(arr * 2)
print(arr // 2)
print(arr**2)
```

### Vectorized Math func

```python
print(np.sqrt(arr))
print(np.floor(arr))
```

```python
print(np.pi)
```

```python
# Excerise

radii = np.array([4, 2, 7])
print(np.pi * radii**2)
```

### Element wise arithmatic

```python
arr1 = np.array([1, 2, 3])
arr2 = np.array([4, 5, 6])
arr3 = np.array([7, 8, 9])

print(arr1 + arr2)
print(arr3 / arr1)
```

### Comparison Operator

```python
scores = np.array([40, 65, 78, 35, 98, 71, 16, 32])

print(scores < 50)

# assign 0 to the students who have failed
scores[scores < 50] = 0
print(scores)
```
