# Broadcasting

Broadcasting allows Numpy to perform operations on array with diff shapes by virtually expanding dimentions so they match the larger arrays shapes

the dimentions have the same size
or 
one of the dimentions has a size of 1
a(1x3) + b(1x3) = result(4x3)


broadcasting rules 
1. They must be equal OR
2. One of them must be exactly 1
 
 If neither of these conditions is met, NumPy throws a ValueError: operands could not be broadcast together.
 
3. Dimension Extension Rule: If one array has fewer dimensions than the other, a dimension of size 1 is prepended (added to the left side) of the smaller array's shape until they have the same number of dimensions.


```python
import numpy as np
```

```python
arr1=np.array([[1,2,3,4]])
arr2=np.array([[1],[2],[3],[4]])

print(arr1.shape)
print(arr2.shape)

print(arr1*arr2)

# (1,4) x (4,1) --- posible 
# (4,1) x (4,4) --- possible 

# (4,1) x (2,4) --- not possible 


```
