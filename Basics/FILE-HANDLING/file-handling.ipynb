# File Handling 

## Reading file 

````python
file=open("fruits.txt","r")
contents = file.read()
file.close()

print(contents)    #consider whole content as a string 
````

````python
# printing raw data 
print(repr(contents))      #for data manipulation we need to split the words 
````

### soritng in alphabetical order 

````python
# spliting the data 
items = contents.split("\n")

for item in items :
    print("--- " + item)
````

## Writing file 

````python

flowers = "Rose\nJasmin\nLily\nDalia\nLotus"
file2 = open("flower.txt","w")
file2.write(flowers)
file2.close()

````

## Appending the string 

````python
more_flower = "\nHibiscus\nMarigold"
file2=open("flower.txt","a")
file2.write(more_flower)
file2.close()
````

#close() method is easy to forgot , so we have package manager 

````python
with open("flower.txt","r") as file3 :
    content =file3.read().split("\n")

for flower in content :
    print(f"- {flower}")
````

````python

#appending the data 
flower_data = ["Orchid","Iris"]
with open("flower.txt","a") as file3 :
    file3.write("\n"+"\n".join(flower_data))   # first \n will  move to the next line after the last element in the txt file , and  "\n".join(flower_data)  will  add \n bw every elements in the list 
 
# reading 
with open("flower.txt","r") as file3 :
    content =file3.read().split("\n")

#printing 
for flower in content :
    print(f"- {flower}")
````

## Problem 1

You've decided to keep a journal. You have your first three entries ready to save:

```python
entries = [
    "Day 1: Started learning Python",
    "Day 2: Went for a run",
    "Day 3: Built my first Python script"
]
```

Write the entries to a file called `journal.txt`, one per line. Then read the file back and print its contents.

**Sample output:**
```
Day 1: Started learning Python
Day 2: Went for a run
Day 3: Built my first Python script
```

````python
entries = [
    "Day 1: Started learning Python",
    "Day 2: Went for a run",
    "Day 3: Built my first Python script"
]

file4 =open("journal.txt",'w')
file4.write("\n".join(entries))    
file4.close()

file4 =open("journal.txt",'r')
data=file4.read()
file4.close()

print(data)
````

## Problem 3

---



You want to look back at your Python-related entries. Using context managers (`with` statements), read `journal.txt` and create a new file called `python_entries.txt` containing only the entries that mention `"Python"`. Then read the new file back and print its contents.

````python
with open("journal.txt","r") as file5 :
    file_entires=file5.read().split("\n")
# print(file_entires)

python_entires=[]
for item in file_entires:
    if "python" in item.lower()  :
        python_entires.append(item)

# print(python_entires)

with open("python_entries.txt","w") as file5 :
    file5.write("\n".join(python_entires))

with open("python_entries.txt","r") as file5 :
    print(file5.read())

````
