### Variable Assignment vs. Context Manager (`with`)

1. **Standard Variable Assignment:**
   ```python
   file = open("abc.txt", "r")
   file.read()
   file.close()
   ```
   * **`file`:** A regular **variable** (file object) created through standard **variable assignment**. You must manually call `close()`.

2. **Context Manager (`with` statement):**
   ```python
   with open("abc.txt", "r") as file:
       file.read()
   ```
   * **`file`:** Technically called the **target variable** (or *context variable*). 
   * **How it works:** Instead of manual assignment, the `as` keyword binds the open file to `file` only for that indented block, automatically closing it once the block finishes.

### Why Forgetting `close()` is a Problem

Forgetting `close()` causes issues in four main situations:

* **Data Loss:** If your program crashes or loses power, unsaved data stuck in memory is lost forever.
* **Empty or Half-Written Reads:** Trying to read the file immediately shows missing data because it hasn't been saved to the disk yet.
* **Program Crashes (`OSError`):** Opening too many files in a loop without closing them hits your computer's limit and crashes the program.
* **Locked Files:** You cannot rename, move, or delete the file on your computer (especially on Windows) while Python is still holding it open.


### Solution - context manager 

````python
with open("Books.txt", "w") as file:
    file.write("Harry Potter and the Philosopher's Stone\n")
    file.write("The Alchemist\n")
    file.write("The Diary of a Young Girl\n")  

print(open("Books.txt", "r").read())
````

#### Context Managers for Databases 

DB- organizer way to store and retrive data 
unlike text files let you search , filter and update data 


##### connect()
##### cursor() --- use to send instructions to the db 

````python
import sqlite3

titles=[
    (1,"The Diary of a Young Girl"),
    (2,"The Count of Monte Cristo"),
    (3,"The Alchemist"),
    (4,"Harry Potter and the Philosopher's Stone")
]

with sqlite3.connect("books.db") as connection :
    cursor = connection.cursor()
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS books(   
        id INTEGER,
        title TEXT
        )
    """)   # IF NOT EXISTS --- if the db not created create one , else keep it 

    for id,title in titles:
        cursor.execute("INSERT INTO books VALUES(?,?)",(id,title))  # (?,?) ---> safe way to insert, collected then inserted all at once  
````

#### splitlines() and split("\n")

```python
text = "apple\nbanana\ncherry\n"

print(text.split("\n"))
# ['apple', 'banana', 'cherry', '']

print(text.splitlines())
# ['apple', 'banana', 'cherry']
```

`split("\n")` adds an empty `''` because the text ends with `\n`; `splitlines()` does not. [python-academy](https://python-academy.org/en/handbook/str.splitlines)

For Windows-style text:

```python
text = "apple\r\nbanana"

print(text.split("\n"))
# ['apple\r', 'banana']

print(text.splitlines())
# ['apple', 'banana']
```

Use `splitlines()` for normal text; use `split("\n")` only when you need the exact `\n` delimiter behavior. [realpython](https://realpython.com/python-split-string/)

````python
text = "0\n22.2\n100\n"

# 1. splitlines() -> Clean list of lines
text.splitlines()   # ['0', '22.2', '100']

# 2. split("\n") -> Leaves an empty string at the end bcoz ends in \n
text.split("\n")    # ['0', '22.2', '100', '']

# 3. split() (no arguments) -> Splits on spaces AND newlines
"hello world\n100".split()  # ['hello', 'world', '100']

# 4. split("") -> Crashes!
text.split("")      # ValueError: empty separator
````
