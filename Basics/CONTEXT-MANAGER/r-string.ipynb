`\` used in escape characters like `\n,\t,\\,\U` etc, 
we can use extra `\` with each `\` in the path like 

```python

```

but it is 
* it is tedious 
* it is error prone

so `r-string ` used 
```python 
file_path=r"D:\Learninggggg\Python_full\Basics\CONTEXT-MANAGER\Books.txt"        ✅

```
