## Problem 1

Use a context manager to write the following city names to a file called `cities.txt`, one city per line:

- Tokyo
- Paris
- New York
- Sydney

Then, use a second context manager to open and read the file, and print how many cities were saved.

Note: use two separate `with` blocks — one to write, one to read. (Combining them on a single line is covered in Problem 2.)

**Sample output**
```
4 cities saved.
```

````python
# TODO write the city names to cities.txt using a context manager

cities=["Tokyo","Paris","New","York","Sydney"]

with open("cities.txt",'w') as file :
    file.write("\n".join(cities))

# TODO read the file using a context manager and print how many cities were saved
with open("cities.txt",'r') as file :
   no= len(file.read().split("\n"))

print(no)
````

## Problem 2

You are given a file `temperatures_c.txt` containing temperatures in Celsius, one per line. Use **two context managers on a single `with` line** to read from `temperatures_c.txt` and write the converted Fahrenheit values to a new file called `temperatures_f.txt`.

The formula to convert Celsius to Fahrenheit is: `F = C * 9/5 + 32`

First, run the cell below to create `temperatures_c.txt`. Then write your solution in the next cell.

**Sample output in `temperatures_f.txt`**
```
32.0
98.6
212.0
72.5
-40.0
```

````python
 # Run this cell first to create the input file
with open("temperatures_c.txt", "w") as f:
    f.write("0\n22.2\n100\n22.5\n-40\n")
````

````python
with (
    open("temperatures_c.txt", "r") as input_file,
    open("temperatures_f.txt", "w") as output_file,
):
   for line in input_file :
      celsius = float(line.strip())
      f=(celsius*9/5)+32
      output_file.write(str(f)+"\n")



````

## Problem 3

Use `sqlite3` and a context manager to:

1. Connect to a database called `store.db`
2. Create a table called `products` using the SQL below. Use `IF NOT EXISTS` so the code can be re-run safely.
   ```sql
   CREATE TABLE IF NOT EXISTS products (
       id INTEGER,
       name TEXT,
       price REAL
   )
   ```
3. Insert the following products using `INSERT INTO products VALUES (?, ?, ?)`:
   - 101, "Wireless Mouse", 25.50
   - 102, "Mechanical Keyboard", 79.99
   - 103, "USB-C Hub", 34.00

**Sample output**

The `products` table in `store.db` should contain:

| id | name | price |
|----|------|-------|
| 101 | Wireless Mouse | 25.50 |
| 102 | Mechanical Keyboard | 79.99 |
| 103 | USB-C Hub | 34.00 |


````python
import sqlite3

data = [
    (101, "Wireless Mouse", 25.50),
    (102, "Mechanical Keyboard", 79.99),
    (103, "USB-C Hub", 34.00),
]

with sqlite3.connect("store.db") as connection:
    cursor = connection.cursor()
    cursor.execute(""" 
        CREATE TABLE IF NOT EXISTS products (
            id INTEGER,
            name TEXT,
            price REAL
        )
""")

    for id,name,price in data :
        cursor.execute("INSERT INTO products VALUES(?,?,?)",(id,name,price))
````
