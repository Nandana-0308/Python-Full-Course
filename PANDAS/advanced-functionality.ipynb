### Shift()

````python
coffee['Yesterday Revenue']=coffee['Revenue'].shift(1)   ## if need revene of 2 days baack --- shift(2)
coffee
````

````python
# coffee['Percentage Change in Revenue']=coffee['Revenue']/coffee['Yesterday Revenue']*100


coffee["Percentage change in Revenue "]=coffee.eval(" `Revenue`/`Yesterday Revenue`*100 ").round(2)

# coffee=coffee.eval(" `Percentage Change in Revenue` = `Revenue`/`Yesterday Revenue`*100 ").round(2)   # when using this for enw variable  no space in variable 


#coffee= not needed if inplace=True  added -- it will directly modify  coffee , otherwise create a copy of coffe and modify


#if inplace=True ---- then round() is not work there ----When you use inplace=True with a calculation inside the string, you cannot add .round() because the entire statement is locked inside the text quotes, and inplace=True returns nothing (None).
# 
# 
#  if inplace= True present ------ coffee.eval(
#     "`Percentage Change in Revenue` = round(`Revenue` / `Yesterday Revenue` * 100, 2)",
#     inplace=True,
# )

coffee
````


Without inplace=True (Making a Copy)

python
coffee = coffee.eval(
    "`Percentage Change in Revenue` = `Revenue` / `Yesterday Revenue` * 100"
)
Use code with caution.
1. Pandas looks at your coffee data.
2. It duplicates the entire dataset in your RAM.
3. It calculates the new column on the duplicate data.
4. It completely overwrites your original coffee variable with the new duplicate.
• The Downside: If your file is huge, duplicating it can slow down your computer or crash your notebook because it uses twice the memory.

✅ With inplace=True (Modifying Directly)

python
coffee.eval(
    "`Percentage Change in Revenue` = `Revenue` / `Yesterday Revenue` * 100",
    inplace=True,
)
Use code with caution.
1. Pandas opens your existing coffee data in memory.
2. It directly tacks the new column onto it without making a copy.
• The Benefit: It is more memory-efficient and keeps your code neat because you don't have to type coffee = coffee... anymore.


````python
coffee.rename(columns={"Percentage change in Revenue ": "Percentage Change In Revenue"}, inplace=True)
coffee
````

````python
coffee['Revenue Rank']=coffee['Revenue'].rank(ascending=True) #lowest Revenue value get lowest
coffee
````

````python
# sort the revenue rank based on the rank 
coffee['Revenue Rank'].sort_values(ascending=False)
````

````python
coffee.sample(6)[['Revenue','Revenue Rank']]
````

````python
latte = coffee[coffee['Coffee Type']=='Latte'].copy()
latte
````

````python
#cumulative sum of latte 
latte['Cumulative Revenue']=latte['Revenue'].cumsum()
latte[['Day','Revenue',"Cumulative Revenue"]]
````

````python
latte['3Day Cumulative Sum']=latte['Revenue'].rolling(3).sum()
latte[['Day','Revenue',"Cumulative Revenue",'3Day Cumulative Sum']]
````
