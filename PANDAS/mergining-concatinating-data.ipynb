# Merging   & Concatinating Data



### 💊 The Pandas `pd.merge()` Capsule

```python
pd.merge(
    left=df1,               # First DataFrame
    right=df2,              # Second DataFrame
    how="inner",            # Join strategy: 'inner', 'left', 'right', 'outer', 'cross'
    on="id",                # Key column name(s) (must exist in BOTH tables)
    left_on=None,           # Use if key column name is unique to the Left table
    right_on=None,          # Use if key column name is unique to the Right table
    left_index=False,       # True = use the Left table's row index as the join key
    right_index=False,      # True = use the Right table's row index as the join key
    sort=False,             # True = sorts the join keys alphabetically/numerically
    suffixes=("_x", "_y"),  # Tuple of strings: renames overlapping non-join columns
    indicator=False,        # True = adds '_merge' column showing where the row came from
    validate=None           # Checks join integrity: '1:1', '1:m', 'm:1', 'm:m'
)
```

#### 🔍 Quick Value Cheat Sheet

* **`how=` Options:** 
  * `"inner"` *(Default)*: Keeps only matching rows found in **both** tables.
  * `"left"`: Keeps **all** from left table + matches from right.
  * `"right"`: Keeps **all** from right table + matches from left.
  * `"outer"`: Keeps **everything** from both tables (fills gaps with `NaN`).
  * `"cross"`: Creates every possible row combination between both tables.
* **`indicator=True` Values:** 
  * Creates a new column named `_merge` containing: `'left_only'`, `'right_only'`, or `'both'`.
* **`validate=` Options:** 
  * `"one_to_one"` or `"1:1"`
  * `"one_to_many"` or `"1:m"`
  * `"many_to_one"` or `"m:1"`
  * `"many_to_many"` or `"m:m"`

### 💊 The Pandas `pd.concat()` Capsule

```python
pd.concat(
    objs=[df1, df2],        # List of DataFrame or Series objects to combine
    axis=0,                 # 0 = Stack rows (top to bottom); 1 = Side-by-side columns (left to right)
    join="outer",           # How to handle indexes on the opposite axis: 'outer' (union) or 'inner' (intersection)
    ignore_index=False,     # True = drops old index numbers and resets them from 0, 1, 2...
    keys=None,              # List/Sequence: Adds a hierarchical/MultiIndex label to identify the source tables
    levels=None,            # Specific levels to use for constructing a MultiIndex (rarely used)
    names=None,             # List: Names for the MultiIndex levels created by 'keys'
    verify_integrity=False, # True = raises an error if the new axis contains duplicate index labels
    sort=False,             # True = sorts non-concatenation axis columns alphabetically if they don't match
    copy=True               # True = copies data; False = reuses memory where possible
)
```

#### 🔍 Quick Direction Cheat Sheet (`axis=`)

* **`axis=0`** *(Default - Row Stacking)*:
  * Glues tables **vertically** (top to bottom).
  * Ideal when you have the exact same columns but new rows of data (e.g., combining January data with February data).
* **`axis=1`** *(Column Pasting)*:
  * Glues tables **horizontally** (left to right).
  * Aligns rows based on their **index labels**.

#### 🔍 Join Options (`join=`)

* **`"outer"`** *(Default)*: Keeps all columns from both tables. If a column exists in `df1` but not `df2`, it fills the missing spots with `NaN`.
* **`"inner"`**: Keeps **only columns** that are present in both `df1` and `df2`. Any unique columns are completely dropped.


````python
pd.concat([fra,rus])
````

### 🎲 Randomly Accessing Rows After Concat

```python
# 1. Quickest Way: Get 5 random rows
df_random = pd.concat([fra, rus], join='inner').sample(n=5)

# 2. Complete Capsule with every parameter:
df_random = pd.concat([fra, rus], join='inner').sample(
    n=None,                 # Number of random rows to return (e.g., n=10)
    frac=0.2,               # Fraction of rows to return (e.g., frac=0.2 returns 20% of the data)
    replace=False,          # True = allow selecting the exact same row multiple times
    weights=None,           # Column name or array to give certain rows a higher chance of being picked
    random_state=42,        # Set an integer seed so you get the EXACT same "random" rows every time you run it
    ignore_index=False      # True = discards original index numbers and resets them to 0, 1, 2...
)
```

#### 💡 3 Common Recipes to Remember

* **Get a specific count:** `.sample(n=1)` gets exactly one random row.
* **Get a percentage:** `.sample(frac=0.5)` randomly grabs half of the dataset.
* **Make it reproducible:** `.sample(n=5, random_state=42)` picks 5 random rows, but ensures that every time you rerun your Jupyter cell, the exact same 5 rows appear (crucial for testing code!).


### ⚔️ Direct Comparison: `pd.merge()` vs `pd.concat()`



| Feature | `pd.merge()` | `pd.concat()` |
| :--- | :--- | :--- |
| **Core Action** | **Relational Joining** (Combines tables horizontally based on shared value keys like a **SQL JOIN / VLOOKUP**). | **Structural Gluing** (Stacks tables vertically or side-by-side like a **Copy/Paste**). |
| **How it Matches Data** | Looks at actual **values inside specific columns** (e.g., matching `id=101` in both tables). | Looks purely at **structural positions** (row index position or column names). |
| **Direction** | **Horizontal only** (adds new columns to a table). | **Both ways** (`axis=0` stacks rows vertically; `axis=1` pastes columns horizontally). |
| **Duplicate Columns** | Handles them using the **`suffixes`** attribute to rename them (e.g., `_x`, `_y`). | Keeps them as-is, resulting in tables with duplicate column names if not careful. |
| **Best Used For** | Linking two different types of datasets together using a **shared ID** (e.g., matching Employees to their Salaries). | Combining multiple datasets with the **same format** together (e.g., stacking Jan, Feb, and Mar sales sheets). |
