````python
coffee=pd.read_csv('./data/coffee.csv')
````

````python
coffee
````

### groupby()

````python
coffee.groupby(['Coffee Type'])['Units Sold'].sum()
````

````python
coffee.groupby(['Coffee Type'])['Units Sold'].mean()
````

### agg()

````python
coffee.groupby(['Coffee Type']).agg({
    'Units Sold' : 'sum'     # putting comma , create more 
})     
````

### pivot Table

````python
pivot=coffee.pivot(columns='Coffee Type',index='Day',values='Units Sold')

pivot
````

````python
pivot.loc['Monday','Latte']
````

````python
pivot.sum()
````

````python
pivot.sum(axis=1)
````

````python
pivot.sum(axis=0)
````

### groupby() combined with datetime operations

````python
bios['born_datetime']=pd.to_datetime(bios['born_date'])
bios.groupby(bios['born_datetime'].dt.year)['name'].count().reset_index().sort_values('name',ascending=False)
````

````python
bios['date_born']=pd.to_datetime(bios['born_date'])
bios['month_born']=bios['date_born'].dt.month
bios['year_born']=bios['date_born'].dt.year
````

````python
bios.groupby([bios['month_born'],bios['year_born']])['name'].count().reset_index().sort_values('name',ascending=False)
````
