## na=False/True/None(Default)


In pandas, the na=False argument is most commonly used inside string methods—such as Series.str.contains()—to specify how missing values (NaN, None, or <NA>) should be treated during data filtering.

By default, string operations on a missing value return NaN. If you try to use that result as a boolean mask to filter a DataFrame, pandas will throw a ValueError because it doesn't know whether to keep or drop the row. Setting na=False explicitly forces pandas to treat those missing values as False


```python


df = pd.DataFrame({'language': ['Python', None, 'Java']})

# Filter using na=False
mask = df['language'].str.contains('Py', na=False)
print(df[mask])
```

```python
# Filter using na=True
mask = df['language'].str.contains('Py', na=True)
print(df[mask])
```

```python
# default value of na  , na=None
mask = df['language'].str.contains('Py') 
print(df[mask]) 
```

```python
df['language'].str.startswith('A', na=False)
```
