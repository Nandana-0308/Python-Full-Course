```python
#adding height category column according to  the height-cm column

bios["height_category"] = bios["height_cm"].apply(
    lambda x: "Short" if x < 165 else ("Average" if x < 185 else "Tall")
)
```

```python
bios['height_category'].tail()
```

```python
# adding column - category acording to the height and weight using function
def categorize_athlete(row):
    if row["height_cm"] < 175 and row["weight_kg"] < 70:
        return "Lightweight"
    elif row["height_cm"] < 185 and row["weight_kg"] <= 80:
        return "Middleweight"
    else:
        return "Heavyweight"


bios["Category"] = bios.apply(categorize_athlete, axis=1)
bios.head()


# Working 
# 1. The Rule (def ...): You create a custom rule that reads an athlete's height and weight, then spits out a label ("Lightweight", "Middleweight", or "Heavyweight").
# 2. The Loop (.apply(..., axis=1)): The .apply() function goes through your dataset row-by-row (because axis=1), feeding each athlete's data into your rule one at a time.
# 3. The Save (bios["Category"] =): It collects all the resulting labels and saves them into a brand-new column named "Category".
# 4. The Preview (.head()): It shows you the first 5 rows of your updated table so you can check the results.
```
