### Accessing specific values - loc and iloc

In pandas, .loc and .iloc are the two primary methods used to select, filter, and slice data from a DataFrame

loc  --- filter by rows and columnns in df  --- loc[rows , columns]

## loc

```python
coffee.loc[0]
```

```python
coffee.loc[0:5]
```

```python
coffee.loc[[0,3,5]]
```

```python
coffee.loc[5:]
```

```python
coffee.loc[:7]
```

```python
coffee.loc[2:5,'Units Sold']
```

```python
coffee.loc[[1,3,4],['Day','Units Sold']]
```

```python
coffee.index=coffee['Day']
coffee
```

```python
coffee.loc['Tuesday']
```

```python
coffee.loc['Monday':'Wednesday','Coffee Type']
```
