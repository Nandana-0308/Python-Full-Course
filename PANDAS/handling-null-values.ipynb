# Handling Null Values 


``` np.nan()
isna()  -- print all NaN contained rows 
notna() -- print all the rows which g=have no NaN values 
fillna()
interpolate()
dropna()

 if chnages needed in the memory --- inplace= True

````python
import pandas as pd
import numpy as np
````

````python
coffee=pd.read_csv('./data/coffee.csv')
````

````python
coffee
````

````python
coffee.loc[[0,3,5,3,12,9],'Units Sold']=np.nan   # loc[0:4,'Units Sold] ---- 0 to 4 rows will have NaN 
coffee
````

````python
coffee.info()
````

````python
#counting the numebr of null values 
coffee.isna().sum()
````

````python
#filling nan  
coffee.fillna(coffee['Units Sold'].mean().round(2))
````

````python
# or 

coffee["Units Sold"] = coffee["Units Sold"].interpolate().bfill()

#  interpolate() didnt change first NaN value , y
# 
# pandas.interpolate() cannot look backward by default
#  solutions 
#  1)   interpolate().bfill() 
#  2)   interpolate(limit_direction="both") 



coffee
````

````python
#drop Nan
coffee.loc[[0,1,2],'Coffee Type'] =np.nan
coffee.loc[[4,6],'Day'] =np.nan

coffee
````

````python
coffee.dropna()
````

````python
coffee.dropna(subset=['Day'])   #specific column with NaN will be removed 
````

````python
coffee[coffee['Coffee Type'].isna()]
````

````python
coffee[coffee['Coffee Type'].notna()] 
````
